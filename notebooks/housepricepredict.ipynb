# 🏠 House Price Prediction With MLflow

This project focuses on:

- 🔍 **Running hyperparameter tuning** while training a regression model for house price prediction  
- 📝 **Logging every hyperparameter and metric** in the MLflow UI for complete traceability  
- 📊 **Comparing the results** of all training runs in the MLflow UI to identify the top-performing configuration  
- ✅ **Registering the best model** in the MLflow Model Registry for versioning and deployment  

---

## 📌 Project Scope

1. **Data Preparation**  
   - Load and clean the house price dataset  
   - Apply feature engineering and selection  
   - Split the data into training and test sets  

2. **Model Training with Hyperparameter Tuning**  
   - Experiment with algorithms such as **Linear Regression**, **Random Forest**, or **XGBoost**  
   - Test multiple hyperparameter combinations  
   - Use cross-validation to measure performance consistently  

3. **MLflow Experiment Tracking**  
   - Define the tracking URI (local or remote server)  
   - Create and organise experiments in MLflow  
   - Log:
     - Hyperparameters (e.g., learning rate, max depth)  
     - Metrics (e.g., RMSE, MAE, R² score)  
     - Artifacts (e.g., plots, feature importance charts)  

4. **Model Comparison**  
   - Use the MLflow UI to review all runs  
   - Sort and filter by key metrics to find the optimal model  
   - Analyse trade-offs between different configurations  

5. **Model Registration**  
   - Add the chosen model to the **MLflow Model Registry**  
   - Assign it to a **stage** (e.g., "Staging", "Production")  
   - Manage version history and promotions  

6. **Deployment Readiness**  
   - Load a model directly from the registry by **version** or **stage**  
   - Prepare the selected model for batch or real-time inference
